```python
import json, sys, pathlib, itertools
import pandas as pd
from statsmodels.stats.contingency_tables import mcnemar
from scipy.stats import chisquare

def load_results(path):
    """Return DataFrame with columns [id, solved]."""
    path = pathlib.Path(path)
    # Accept either a JSON array or JSON-lines
    try:
        data = json.loads(path.read_text())
    except json.JSONDecodeError:
        data = [json.loads(line) for line in path.read_text().splitlines() if line.strip()]
    return pd.DataFrame(
        [(item["id"], bool(item["solved"])) for item in data],
        columns=["id", "solved"]
    )

def generate_results(path1, path2):
    

    dfA = load_results(path1).rename(columns={"solved": "JSON"})
    dfB = load_results(path2).rename(columns={"solved": "XML"})
    # --- McNemar test (paired models) ---
    df = pd.merge(dfA, dfB, on="id", how="inner")
    table = pd.crosstab(df["JSON"], df["XML"])
    # Ensure full 2×2 shape
    table = table.reindex(index=[False, True], columns=[False, True], fill_value=0)

    print("Contingency table (JSON rows × XML columns):")
    print(table, "\n")

    result = mcnemar(table.values, exact=False, correction=True)
    print("McNemar χ² =", result.statistic, "   p-value =", result.pvalue)

```

```python
import os
directory = "/home/js2042/Documents/projects/ZebraLogicBench/result_dirs_parsed/zebra-grid-mini/json"
directory2 = "/home/js2042/Documents/projects/ZebraLogicBench/result_dirs_parsed/zebra-grid-mini/xml"
for filename in os.listdir(directory):
    print(filename)
    path1 = os.path.join(directory, filename)
    path2 = os.path.join(directory2, filename)
    generate_results(path1, path2)
    print("\n\n=========================\n\n")
    
# generate_results(path1, path2)
```
