```python
import json
```

```python
def load_json(filename):
    with open(filename, 'r') as f:
        return json.load(f)
    
def save_file(filename, data):
    with open(filename, 'w') as f:
        json.dump(data, f, indent=4)
```

```python
f1 = 'result_dirs/zebra-grid/json/tmp_Qwen25-14B-Coder-Instruct/Qwen25-14B-Coder-Instruct.0-250.json'
f2 = 'result_dirs/zebra-grid/json/tmp_Qwen25-14B-Coder-Instruct/Qwen25-14B-Coder-Instruct.250-500.json'
f3 = 'result_dirs/zebra-grid/json/tmp_Qwen25-14B-Coder-Instruct/Qwen25-14B-Coder-Instruct.500-750.json'
f4 = 'result_dirs/zebra-grid/json/tmp_Qwen25-14B-Coder-Instruct/Qwen25-14B-Coder-Instruct.750-1000.json'

data1 = load_json(f1)
data2 = load_json(f2)
data3 = load_json(f3)
data4 = load_json(f4)

print(type(data1))
print(len(data1))
data = data1 + data2 + data3 + data4
print(len(data))

save_file('result_dirs/zebra-grid/json/Qwen25-14B-Instruct.json', data)
```
